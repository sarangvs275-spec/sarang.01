```python
#6)
a=int(input('enter the day(1-7)'))
match a:
    case 1:
        print('monday')
    case 2:
        print('tuesday')
    case 3:
        print('wednesday')
    case 4:
        print('thursday')
    case 5:
        print('friday')
    case 6:
        print('saturday')
    case 7:
        print('sunday')
```

```python
#7)
a=int(input('enter the value'))
b=int(input('enter the second value'))
op = input ('enter the operators(+-*/)')
match op:
    case '+':
        print(a+b)
    case '-':
        print(a-b)
    case '*':
        print(a*b)
    case '/':
        print(a/b)
```

```python
# #8)
grade = input("Enter grade (A, B, C, D, F): ").upper()

match grade:
    case "A":
        print("Excellent")
    case "B":
        print("Very Good")
    case "C":
        print("Good")
    case "D":
        print("Pass")
    case "F":
        print("Fail")
    case _:
        print("Invalid Grade")
```

```python
#9)
signal = input("Enter traffic signal (Red, Yellow, Green): ").lower()

match signal:
    case "red":
        print("Stop")
    case "yellow":
        print("Get Ready")
    case "green":
        print("Go")
    case _:
        print("Invalid Signal")
```
